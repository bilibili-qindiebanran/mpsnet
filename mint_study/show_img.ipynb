## 两种读取并显示图片的方法

```python
from PIL import Image

# 读取图像
img = Image.open('../Images/200_samples/ntust-hp/0003/01.png')

# 显示图像
img.show()
```

```python
import cv2

# 读取图像 (BGR格式)
img = cv2.imread('../Images/200_samples/ntust-hp/0003/01.png')

# 会弹出一个图形化窗口，jupyter可能无法正常显示
# # 显示图像
# cv2.imshow('circle', img2)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

from matplotlib import pyplot as plt

# img_rgb = img2[:, :, ::-1] # 转为 RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # 转为 RGB
plt.figure(figsize=(10, 10))
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```
